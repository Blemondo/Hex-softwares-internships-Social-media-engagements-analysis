# Project 01 – Social Media Engagement Analysis

Analyze 600 social media posts to understand which content performs best. The dataset is synthetic and designed for educational use.


## 1. Import libraries


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
```

## 2. Load and inspect the dataset


```python
df=pd.read_csv('social_media_engagement_600.csv')
print(df.shape)
display(df.head())
display(df.info())
```

## 3. Clean the data


```python
print(df.isnull().sum())
print('Duplicates:',df.duplicated().sum())
df['Comments']=df['Comments'].fillna(df['Comments'].median())
df['Date']=pd.to_datetime(df['Date'])
```

## 4. Calculate engagement metrics


```python
df['Engagements']=df[['Likes','Comments','Shares','Saves']].sum(axis=1)
df['Engagement_Rate']=df['Engagements']/df['Reach']*100
display(df.head())
```

## 5. Analysis


```python
content=df.groupby('Content_Type').agg(Posts=('Content_Type','size'),Avg_Engagement=('Engagements','mean'),Avg_Engagement_Rate=('Engagement_Rate','mean')).sort_values('Avg_Engagement_Rate',ascending=False)
platform=df.groupby('Platform')['Engagements'].mean().sort_values(ascending=False)
print(content)
print(platform)
```

## 6. Visualization 1 — Engagement rate by content type


```python
content['Avg_Engagement_Rate'].plot(kind='bar',figsize=(9,5)); plt.title('Average Engagement Rate by Content Type'); plt.ylabel('Engagement Rate (%)'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

## 7. Visualization 2 — Average engagement by platform


```python
platform.plot(kind='bar',figsize=(9,5)); plt.title('Average Engagements by Platform'); plt.ylabel('Average Engagements'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

## 8. Visualization 3 — Engagement-rate distribution


```python
plt.figure(figsize=(9,5)); sns.histplot(df['Engagement_Rate'],bins=30,kde=True); plt.title('Distribution of Engagement Rates'); plt.xlabel('Engagement Rate (%)'); plt.tight_layout(); plt.show()
```

## 9. Visualization 4 — Reach vs engagements


```python
plt.figure(figsize=(9,5)); sns.scatterplot(data=df,x='Reach',y='Engagements',hue='Content_Type'); plt.title('Reach vs Engagements'); plt.tight_layout(); plt.show()
```

## 10. Visualization 5 — Monthly engagement trend


```python
df['Month']=df['Date'].dt.to_period('M').astype(str); monthly=df.groupby('Month')['Engagements'].sum(); plt.figure(figsize=(11,5)); monthly.plot(marker='o'); plt.title('Total Engagements Over Time'); plt.xlabel('Month'); plt.ylabel('Engagements'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
